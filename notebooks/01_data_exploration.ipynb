```python
combined_df["Match_ID"] = (
    combined_df["Y/M/D"].astype(str)
    + "_"
    + combined_df.apply(
        lambda row: "_".join(
            sorted([row["Team"], row["VS_Team"]])
        ),
        axis=1
    )
)
```

```python
combined_df = combined_df.sort_values(
    ["Y/M/D", "Match_ID"]
).reset_index(drop=True)
```

```python
combined_df.to_clipboard(excel=True, sep="\t", index=False)
```

```python
combined_df["Won"] = (
    combined_df["SET_Won"] > combined_df["SET_Lost"]
).astype(int)
```

```python
point_columns = [
    "Point1",
    "Point2",
    "Point3",
    "Point4",
    "Point5",
    "VS_Point1",
    "VS_Point2",
    "VS_Point3",
    "VS_Point4",
    "VS_Point5"
]

combined_df[point_columns] = combined_df[point_columns].replace("-", "0")
combined_df[point_columns] = combined_df[point_columns].fillna(0)
```

```python
combined_df = combined_df.drop(columns=["SET"])
    
```

```python

import numpy as np

combined_df["Attack_Efficiency"] = np.where(
    combined_df["ATTACK_Attempts"] > 0,
    (
        combined_df["ATTACK_Point"]
        - combined_df["ATTACK_Errors"]
    ) / combined_df["ATTACK_Attempts"],
    np.nan
)


combined_df["SERVE_Efficiency"] = np.where(
    combined_df["ATTACK_Attempts"] > 0,
    (
        combined_df["SERVE_Point"]
        - combined_df["SERVE_Errors"]
    ) / combined_df["SERVE_Attempts"],
    np.nan
)


combined_df["RECEPTION_Efficiency"] = np.where(
    combined_df["RECEPTION_Attempts"] > 0,
    (
        combined_df["RECEPTION_Successful"]
        - combined_df["RECEPTION_Errors"]
    ) / combined_df["RECEPTION_Attempts"],
    np.nan
)

combined_df["BLOCK_Efficiency"] = np.where(
    combined_df["BLOCK_Touches"] > 0,
    (
        combined_df["BLOCK_Point"]
        - combined_df["BLOCK_Errors"]
    ) / combined_df["BLOCK_Touches"],
    np.nan
)

combined_df["DIG_Efficiency"] = np.where(
    combined_df["DIG_Total"] > 0,
    (
        combined_df["DIG_Digs"]
        - combined_df["DIG_Errors"]
    ) / combined_df["DIG_Total"],
    np.nan
)


combined_df["Attack_Efficiency"] = (
    (combined_df["ATTACK_Point"] - combined_df["ATTACK_Errors"])
    / combined_df["ATTACK_Attempts"]
)

combined_df["Serve_Efficiency"] = (
    (combined_df["SERVE_Point"] - combined_df["SERVE_Errors"])
    / combined_df["SERVE_Attempts"]
)

combined_df["Reception_Efficiency"] = (
    (
        combined_df["RECEPTION_Successful"]
        - combined_df["RECEPTION_Errors"]
    )
    / combined_df["RECEPTION_Attempts"]
)

combined_df["BLOCK_Efficiency"] = (
    (combined_df["BLOCK_Point"] - combined_df["BLOCK_Errors"])
    / combined_df["BLOCK_Touches"]
)

combined_df["DIG_Efficiency"] = (
    (combined_df["DIG_Digs"] - combined_df["DIG_Errors"])
    / combined_df["DIG_Total"]
)


```
