```python
import pandas as pd
import numpy as np
import os



import pandas as pd

from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import RidgeClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)


df = pd.read_csv("/Users/zu/Desktop/projects/vballPredictor/data/features/vnl_features.csv")
df.head()

df["Y/M/D"] = pd.to_datetime(df["Y/M/D"], format="%Y-%m-%d")

feature_columns = [
    "Prior_Win_Rate_Diff",
    "Recent_Form_Diff",
    "Prior_ATTACK_Efficiency_Diff",
    "Prior_SERVE_Efficiency_Diff",
    "Prior_RECEPTION_Efficiency_Diff",
    "Prior_BLOCK_Efficiency_Diff",
    "Prior_DIG_Error_Rate_Diff",
    "Prior_SET_Error_Rate_Diff"
]

x = df[feature_columns]
y = df["Won"]

training_df = df[df["Y/M/D"].dt.year < 2025]

testing_df = df[df["Y/M/D"].dt.year == 2025]

X_train = training_df[feature_columns]
y_train = training_df["Won"]

X_test = testing_df[feature_columns]
y_test = testing_df["Won"]


scaler = StandardScaler()
scaler.fit(X_train)

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model1 = RidgeClassifier(alpha=1.0)
model1.fit(X_train_scaled, y_train)
y_pred = model1.predict(X_test_scaled)

results = testing_df[
    ["Y/M/D", "Team", "VS_Team", "Won"]
].copy()

results["Predicted"] = y_pred

accuracy = accuracy_score(
    y_test,
    y_pred
)

print(f"Accuracy: {accuracy:.3f}")

print(
    classification_report(
        y_test,
        y_pred
    )
)

print(
    f"Accuracy: {accuracy_score(y_test, y_pred):.3f}"
)

print(
    f"Precision: {precision_score(y_test, y_pred):.3f}"
)

print(
    f"Recall: {recall_score(y_test, y_pred):.3f}"
)

print(
    f"F1 Score: {f1_score(y_test, y_pred):.3f}"
)

cm = confusion_matrix(
    y_test,
    y_pred
)

print(cm)

feature_importance = pd.DataFrame({
    "Feature": feature_columns,
    "Coefficient": model1.coef_[0]
})

feature_importance = feature_importance.sort_values(
    "Coefficient",
    ascending=False
)

print(feature_importance)

results.head(10)

```
